```python
from crewai import Agent, Task, Crew, Process
```

```python
# Define Customer Support Agent
customer_support_agent = Agent(
    role='Customer Support',
    goal='Provide accurate and timely responses to customer inquiries',
    backstory='An AI designed to handle customer support tickets efficiently and courteously.',
    tools=[],  # Add tools if applicable
    llm='gpt-4o-mini',
    verbose=True
)

customer_inquiry = 'I bought an item and I dont like it. I want to return it. What is your return policy for this case?'

support_task = Task(
    description=f"Respond to the following customer inquiry: {customer_inquiry}",
    expected_output='A detailed explanation of the return policy',
    agent=customer_support_agent
)
```

```python
crew = Crew(
    agents=[customer_support_agent],
    tasks=[support_task],
    process=Process.sequential
)

result = crew.kickoff()
print(result.raw)
```
